```python
import sys
import io
class SuppressPrints:  # Context manager to silence function output
    def __enter__(self):
        self._original_stdout = sys.stdout
        sys.stdout = io.StringIO()
    def __exit__(self, exc_type, exc_val, exc_tb):
        sys.stdout = self._original_stdout
```

```python
import os
from PIL import Image
import torch
import torchvision.transforms as transforms
from torchvision.utils import save_image


# Set up device and delta (your adversarial perturbation)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Directories for input images and saving adversarial images
input_dir = '/content/test/images'
output_dir = '/content/adversarial_test_images_square'
os.makedirs(output_dir, exist_ok=True)

i=0
# Loop through images in the input directory
for filename in os.listdir(input_dir):
    if filename.lower().endswith(('.png', '.jpg', '.jpeg')):
        image_path = os.path.join(input_dir, filename)
        with SuppressPrints():
          adv = square_attack_detector(
          image_path=image_path,
          model_path=model_path,
          epsilon=0.05,
          patch_size=256,
          num_iterations=200,
          conf_threshold=0.25,
          device=device)

          if adv is None:
            continue

        # Save the adversarial image as a PNG file using torchvision's save_image
        # Modify filename to ensure it has a .png extension
        base_filename = os.path.splitext(filename)[0]
        output_path = os.path.join(output_dir, f"{base_filename}.png")
        save_image(adv, output_path)
        i = i+1
        if i%50 == 0:
          print(i)
```
